```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
iris = datasets.load_iris()
```

```python
iris_data = pd.DataFrame(iris.data)
```

```python
iris_data.columns = iris.feature_names
```

```python
iris_data
```

## check this Lance–Williams update formula

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(iris_data)
```

```python
X_scaled
```

```python
# Apply the PCA
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
```

```python
X_pca = pca.fit_transform(X_scaled)
```

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1], c= iris.target)
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
```

```python
#Agglomerative Hierarchical clustering
from sklearn.cluster import AgglomerativeClustering
ac = AgglomerativeClustering(n_clusters=3, metric = 'euclidean', linkage='ward')

ac.fit(X_pca)
```

```python
ac.labels_
```

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1], c= ac.labels_)
```

```python
iris.target
```

```python
# Silhouette score
from sklearn.metrics import silhouette_score

silhouette_coefficients = []
for k in range(2, 12):
  ac = AgglomerativeClustering(n_clusters=k, metric = 'euclidean', linkage='ward')
  ac.fit(X_pca)
  score = silhouette_score(X_pca, ac.labels_)
  silhouette_coefficients.append(score)
```

```python
plt.plot(range(2,12),silhouette_coefficients)
```
