```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import SparseCategoricalCrossentropy
```

```python
mnist = tf.keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train, X_test = X_train / 255.0, X_test / 255.0
```

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation="softmax")
])
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10)
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=2)
print('\nTest accuracy:', test_acc)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
import numpy as np
top2_idx = np.argsort(y_pred[0])[-2:][::-1]      # indices of top 2 probs
top2_prob = y_pred[0][top2_idx]                  # their probabilities

for digit, prob in zip(top2_idx, top2_prob):
    print(f"Digit {digit} → Probability {prob:.4f}")
```

```python
top2_idx_all = np.argsort(y_pred, axis=1)[:, -2:][:, ::-1]   # top 2 indices
top2_prob_all = np.take_along_axis(y_pred, top2_idx_all, axis=1)  # top 2 probs
```

```python
top2_idx_all
```

```python
import pandas as pd
result_df = pd.DataFrame({
    'y_test': y_test,
    'top1_digit': top2_idx_all[:, 0],
    'top1_prob': top2_prob_all[:, 0],
    'top2_digit': top2_idx_all[:, 1],
    'top2_prob': top2_prob_all[:, 1]
})
```

```python
result_df
```

```python
result_df.to_csv("result.csv")
```

## Lets apply CNN

```python
X_train.shape
```

```python
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

model2 = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),

    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
model2.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
model2.fit(X_train, y_train, epochs=2, validation_data=(X_test, y_test))
```

```python
test_loss2, test_acc2 = model2.evaluate(X_test, y_test, verbose=2)
print('\nTest accuracy:', test_acc2)
```

```python
y_pred2 = model2.predict(X_test)
```
