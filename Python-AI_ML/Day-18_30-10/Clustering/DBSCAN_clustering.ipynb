```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
import numpy as np
```

```python
X, y = make_moons(n_samples= 1000, noise=0.05)
```

```python
plt.scatter(X[:, 1], X[:, 0])
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X.shape
```

```python
dbscan = DBSCAN(eps = 0.15)
```

```python
dbscan.fit(X_scaled)
```

```python
dbscan.labels_
```

```python
plt.scatter(X_scaled[:, 1], X_scaled[:, 0], c=dbscan.labels_)
```

Now for points not included in clusters, including min points

```python
dbscan2 = DBSCAN(eps = 0.15, min_samples=10)
```

```python
dbscan2.fit(X_scaled)
```

```python
plt.scatter(X_scaled[:, 1], X_scaled[:, 0], c=dbscan2.labels_)
```

```python
dbscan2.labels_
```

```python
unique, counts = np.unique(dbscan2.labels_, return_counts=True)
result = dict(zip(unique, counts))
print(result)
```
