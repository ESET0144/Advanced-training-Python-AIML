```python
import pandas as pd
df = pd.read_csv("healthcare.csv")
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.scatter(df.iloc[:,0], df.iloc[:,1])
```

```python
from sklearn.ensemble import IsolationForest
clf = IsolationForest(contamination=0.05) # contamination - accepted outliers upto a point
clf.fit(df)
predictions = clf.predict(df)
```

```python
predictions
```

```python
x= df.values
```

```python
import numpy as np
index = np.where(predictions<0)
index
```

```python
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.scatter(x[index,0],x[index,1],edgecolors='r')
```

## Trying isolation forest on moons to check is it more suitable for spheres only or general

```python
from sklearn.datasets import make_moons
```

```python
X, y = make_moons(n_samples= 1000, noise=0.1)
```

```python
plt.scatter(X[:, 1], X[:, 0])
```

```python
clf = IsolationForest(contamination=0.05) # contamination - accepted outliers upto a point
clf.fit(X)
predictions = clf.predict(X)
```

```python
import numpy as np
index = np.where(predictions<0)
index
```

```python
plt.scatter(X[:,0],X[:,1])
plt.scatter(X[index,0],X[index,1],edgecolors='r')
```
